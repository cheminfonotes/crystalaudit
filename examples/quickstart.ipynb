# crystalaudit — Quickstart

Automated crystallographic validation & anomaly attribution.
Existing tools *flag* geometry outliers; **crystalaudit explains why** —
biological strain, crystal packing, or refinement error.

Works the same locally and in **Google Colab**.

## 1 · Install (Colab / local — one line)

```python
# In Colab or a fresh environment, install straight from GitHub:
!pip install -q git+https://github.com/cheminfonotes/crystalaudit.git
# (local dev: `pip install -e .` from the repo root)
```

## 2 · Analyze a structure in **one line**\nGive it a 4-char PDB ID — it downloads from RCSB automatically.

```python
import crystalaudit as ca

ca.analyze("1UBQ").show()      # inline plot + table
```

## 3 · Rich auto-display
In a notebook, a **bare expression** renders an HTML summary (chips + table) —
no `.show()` needed.

```python
ca.analyze("1UBQ")
```

## 4 · Get the results as a DataFrame

```python
res = ca.analyze("1UBQ")
res.report.head()
```

## 5 · From a local / uploaded file
Any `.pdb` or `.cif` path works. The repo ships a tiny `sample_structure.pdb`
for offline testing.

```python
res = ca.analyze("sample_structure.pdb")   # or any path you have
print(res.counts)
res.report.head()
```

## 6 · Tune & export

```python
res = ca.analyze("1UBQ", z_outlier=4.0, contact_radius=4.5)
res.to_csv("ubq_report.csv")
res.to_html("ubq_report.html")
res.summary()
```

## 7 · Just the plot
`.plot()` returns a matplotlib `Figure` you can save or restyle.

```python
fig = ca.analyze("1UBQ").plot()
fig.savefig("overview.png", dpi=130)
```

---
### What you get back — the `Result` object

| Member | What |
|---|---|
| `.report` | DataFrame of attributed outliers |
| `.counts` | `{'biological': n, 'packing': n, 'error': n}` |
| `.plot()` / `.show()` | figure / inline figure+table |
| `.to_csv()` / `.to_html()` | export |
| `.summary()` | headline numbers |

**Command line:** `crystalaudit 1UBQ --csv out.csv --png fig.png`
